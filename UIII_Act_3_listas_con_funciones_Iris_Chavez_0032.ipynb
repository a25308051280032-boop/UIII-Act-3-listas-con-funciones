11. Plomero

```python
# Base de datos temporal (Lista de diccionarios)
plomeros = []

def agregar_plomero():
    print("\n--- Registro de Nuevo Plomero ---")
    id_p = int(input("Ingrese ID: "))
    nombre = input("Nombre: ")
    experiencia = int(input("Años de experiencia: "))
    costo_hora = float(input("Costo por hora: "))

    # Operador lógico y relacional para validar "Plomero Senior"
    es_senior = experiencia >= 10 and costo_hora > 50

    nuevo = {
        "id": id_p,
        "nombre": nombre,
        "experiencia": experiencia,
        "costo": costo_hora,
        "senior": es_senior
    }
    plomeros.append(nuevo)
    print("Plomero agregado con éxito.")

def listar_plomeros():
    print("\n--- Lista de Plomeros ---")
    # Operador relacional: verificar si la lista no está vacía
    if len(plomeros) == 0:
        print("No hay registros.")
    else:
        for p in plomeros:
            print(f"ID: {p['id']} | Nombre: {p['nombre']} | Exp: {p['experiencia']} años | Senior: {p['senior']}")

def buscar_plomero():
    busqueda = int(input("\nIngrese el ID a buscar: "))
    encontrado = False
    for p in plomeros:
        # Operador relacional de igualdad
        if p['id'] == busqueda:
            print(f"Resultado: {p}")
            encontrado = True
            break
    if not encontrado: # Operador lógico NOT
        print("Plomero no encontrado.")

def modificar_plomero():
    id_m = int(input("\nID del plomero a modificar: "))
    for p in plomeros:
        if p['id'] == id_m:
            p['nombre'] = input("Nuevo nombre: ")
            p['costo'] = float(input("Nuevo costo por hora: "))
            # Operador aritmético: ajuste de costo por inflación (ejemplo 10%)
            p['costo'] = p['costo'] * 1.10
            print("Datos actualizados (se aplicó un +10% por ajuste de temporada).")
            return
    print("ID no existe.")

def borrar_plomero():
    id_b = int(input("\nID del plomero a eliminar: "))
    for i in range(len(plomeros)):
        if plomeros[i]['id'] == id_b:
            # Borrado por índice
            del plomeros[i]
            print("Registro eliminado.")
            return
    print("No se encontró el ID.")

# --- Menú Principal ---
def menu():
    while True:
        print("\n1. Agregar | 2. Listar | 3. Buscar | 4. Modificar | 5. Borrar | 6. Salir")
        opcion = input("Seleccione: ")
        if opcion == '1': agregar_plomero()
        elif opcion == '2': listar_plomeros()
        elif opcion == '3': buscar_plomero()
        elif opcion == '4': modificar_plomero()
        elif opcion == '5': borrar_plomero()
        elif opcion == '6': break
        else: print("Opción inválida.")

menu()
print("Prpgrama realizado por Iris Chavez NL = 11")
```
